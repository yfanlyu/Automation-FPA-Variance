```python
from pathlib import Path
import pandas as pd
import numpy as np
```

```python
OUT = Path("/Users/cici/Desktop/Learning/python_work/Project_idea/fpna_variance_project/src/data_generation/data/output")
```

```python
def add_anomaly_score():
    variance = pd.read_csv(OUT/"actual_vs_budget_variance.csv", parse_dates=["date"])
    

    x = variance.sort_values(["region_id", "account_id", "date"])
    x["rolling_mean"] = x.groupby(["region_id", "account_id"])["variance"].transform(
        lambda s: s.shift(1).rolling(6, min_periods=3).mean())
    x["rolling_std"] = x.groupby(["region_id", "account_id"])["variance"].transform(
        lambda s: s.shift(1).rolling(6, min_periods=3).std())
    x["z_score"] = (x.variance - x.rolling_mean) / x.rolling_std.replace(0,pd.NA)
    x["anomaly"] = x.z_score.abs() > 3

    x.to_csv(OUT/f"anomaly.csv")
    
    return x
```

```python
add_anomaly_score()
```
